```python
# 어떤 마을에 100개의 집이 일렬로 나란히 있습니다.
# 각 집에는 스위치가 하나씩 있고, 처음에는 모두 꺼져 있습니다(OFF).

# 1번부터 100번까지의 집 앞을 1번 학생부터 100번 학생까지 각각 한 번씩 지나갑니다.

# 1번 학생은 모든 집의 스위치를 누릅니다(1, 2, 3, ..., 100번 집).

# 2번 학생은 2의 배수 집(2, 4, 6, ..., 100번 집)의 스위치를 누릅니다.

# 3번 학생은 3의 배수 집(3, 6, 9, ..., 99번 집)의 스위치를 누릅니다.

# ...

# 100번 학생은 100번 집의 스위치만 누릅니다.

# 각 학생이 지나갈 때마다 해당 집의 스위치는 ON/OFF가 바뀝니다.

# 질문
# 마지막에 스위치가 켜져 있는 집의 번호는 모두 몇 번일까요?
```

```python
n = 100

memo = {}
for i in range(1, n + 1) :
    count = 0 
    for j in range(1, i + 1) :
        if i % j == 0 :
            count += 1
            memo[i] = count

print(memo)

result = []
for k, v in memo.items() :
    if v % 2 == 1 :
        result.append(k)

print(f'불이 켜져 있는 집은 {result}')
```
