```python
# 1. 숫자 문자열 판별하기
# 사용자로부터 문자열을 입력받아, 해당 문자열이 정수로만 이루어져 있는지 판별하는 프로그램을 작성하세요.

user = 'fjskld9080'

print(user.isdigit())
print(user.isalpha())
print(user.isnumeric())
print(user.isdecimal())
print(user.isalnum())
```

```python
# 2. 알파벳과 숫자 혼합 판별하기
# 아래의 문자열 리스트에서 알파벳과 숫자가 모두 포함된 문자열만 골라 새로운 리스트로 만드세요.

lst = ["abc123", "1234", "abcd", "a1b2", "5678", "xyz"]

def new_list(data) :
    result = []
    for i in data :   
        if i.isdigit() == False and i.isalpha() == False :
            result.append(i)

    return result

print(new_list(lst))

def alphanum_list(data) :
    result = []
    for i in data :
        if not i.isdigit() and not i.isalpha() :    # not 을 사요하는 것이 False 를 사용하는 것 보다 python 적이다
            result.append(i)
    return result
print(alphanum_list(lst))

print(list(filter(lambda x : x.isdigit() == False and x.isalpha() == False , lst)))
print(list(filter(lambda x : not x.isdigit() and not x.isalpha(), lst)))
```

```python
# 3. 부분 문자열 포함 여부 확인
# 아래의 문자열 리스트에서 "py"라는 부분 문자열이 포함된 항목이 있는지 확인하여,
# 있으면 "포함됨", 없으면 "포함되지 않음"을 출력하세요.

lst = ["java", "python", "c++", "javascript"]

def sub_words(str, data) :
    result = {}
    for i in data :
        if (str in i for i in data) == True :
            result[i] = '포함됨'
        else :
            result[i] = '포함안됨'
    return result

print(sub_words('py', lst))    


print(dict(map(lambda x : (x,'포함됨') if 'py' in x  in lst else (x,'포함안됨'), lst)))
```

```python
# 4. 리스트에 값이 모두 있는지 확인
# 아래의 리스트에서 모든 값이 0이 아닌지 확인하는 코드를 작성하세요.
# 모두 0이 아니면 "모두 0이 아님"을, 하나라도 0이면 "0이 포함됨"을 출력하세요.

lst = [1, 2, 3, 4, 5, 0]

def compare_item(data) :
    if 0 in data :
        return '0 이 포함 됨'
    else : 
        return '0 이 포함 안됨'

print(compare_item(lst))

print('포함됨' if 0 in lst else '포함안됨')
```
