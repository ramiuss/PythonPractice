```python
memo = {}

def fibo(n) : 
    if n in memo :
        return memo[n]
    
    if n ==1 or n ==2 :
        return 1
    
    else :
        temp = (fibo(n-1) + fibo(n-2))
        memo[n] = temp
        return temp
    
print(fibo(10), memo)
print(fibo(20), memo)
```

```python
memo = {1:1, 2:1}

def fibo(n) :
    if n in memo :
        return memo[n]
    else :
        temp = fibo(n-1) + fibo(n-2) 
        memo[n] = temp
        return temp
    
print(fibo(10))
print(fibo(50))
```

```python
data = [[1, 2, 3, ], [4, [5, 6]], 7, [8, 9]]

def flatten(data) :
    result = []
    for i in data :
        if type(i) == list :
            result.extend(flatten(i))
        else : 
            result.append(i)
    return result

print(flatten(data))
```

```python
a = 'aaabbcddddeeaaa'

def compress_string(a):
    # if not a:
    #     return ""
    result = []
    curr_char = a[0]
    count = 0
    for c in a:
        if c == curr_char:
            count += 1
        else:
            result.append(curr_char + str(count))
            curr_char = c
            count = 1
    # 마지막 문자 처리
    result.append(curr_char + str(count))
    return ''.join(result)

# 예시 실행
print(compress_string(a))  # a3b2c4d1a2

```
