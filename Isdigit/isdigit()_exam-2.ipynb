```python
# 1. 문자열이 모두 대문자인지 확인하기
# 사용자로부터 문자열을 입력받아, 해당 문자열이 모두 대문자인지 여부를 출력하는 프로그램을 작성하세요.

user = 'fjsdlfja'

def capital(data) :
    if data.isupper() :
        return True
    else :
        return False
    
print(capital('fjsdo'))
print(capital('FJDSLFS'))
print(capital('FJSLAfjdls'))
```

```python
# 2. 공백으로만 이루어진 문자열 판별하기
# 아래 문자열 리스트에서 공백 문자(스페이스, 탭, 줄바꿈 등)로만 이루어진 문자열을 찾아 출력하세요.

lst = ["  ", "\t", "abc", "\n", "   a"]

def space_words(data) :
    result = []
    for i in data :
        if i.isspace() :
            result.append(i)
    return result
print(space_words(lst))

print(list(filter(lambda x :  x.isspace(), lst)))
```

```python
# 3. 리스트 내 문자열 중 접두사 검사
# 아래 리스트에서 "pre"로 시작하는 문자열만 골라 새로운 리스트로 만드세요.

lst = ["prefix", "presentation", "apple", "prelude", "banana"]

def prefix_word(data) :
    result = []
    for i in data :
        if i.startswith('pre') :
            result.append(i)
    return result
print(prefix_word(lst))
```

```python
# 4. 리스트에 특정 값이 포함되어 있는지 확인
# 아래 리스트에서 숫자 10이 포함되어 있으면 "10 있음", 없으면 "10 없음"을 출력하세요.

lst = [5, 7, 11, 12, 15, 10]

def num_numbers(num, data) :
    if num in data :
        return '10 있음'
    else :
        return '10 없음'
print(num_numbers(10, lst))

print('10 있음' if 10 in lst else '10 없음')
```

```python
# 5. 리스트 내 문자열이 모두 알파벳인지 확인
# 아래 리스트의 모든 문자열이 알파벳 문자로만 이루어져 있는지 확인하는 코드를 작성하세요.
# 모두 알파벳이면 "모두 알파벳", 아니면 "알파벳 아닌 문자열 있음"을 출력하세요.

lst = ["hello", "world", "Python3", "code", '2345' ]

def alpha_words(data) :
    for i in data :
        if not i.isalpha() :
            return '알파벳이 아닌 문자 있음음'
        
    return '모두 알파벳'
print(alpha_words(lst))
```
