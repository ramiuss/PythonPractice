```python
# 1. 아나그램(anagram) 판별 함수
# 두 개의 문자열을 입력받아, 두 문자열이 아나그램(철자만 다르고 구성은 같은 단어)이면 True, 아니면 False를 반환하는 함수를 작성하세요.

def anagrom(string1, string2) :
    for i in string1 :
        if i not in string2 :
            return False
        else :
            return True
        
print(anagrom('listen', 'silent'))
print(anagrom('hellow', 'world'))
```

```python
# 2. 회문(palindrome) 판별 함수
# 문자열을 입력받아, 앞뒤가 같은 회문이면 True, 아니면 False를 반환하는 함수를 작성하세요.

def palindrome(string) :
    if string == string[::-1] :
        return True
    else : 
        return False

print(palindrome('abba'))
print(palindrome('racecar'))
print(palindrome('python'))
```

```python
# 3. 딕셔너리에서 값이 가장 큰 키 찾기
# 딕셔너리를 입력받아, 값이 가장 큰 키를 반환하는 함수를 작성하세요.

def dictionary(dict) :
    max = 0 
    for k, v in dict.items() :
        if dict[k] > max :
            max = dict[k]
        else :
            continue
    return k, dict[k]

a = {1 : 10, 2 : 20, 3 : 30, 4 : 40, 5 : 50}

print(dictionary(a))
```

```python
# 4. 중첩 리스트(flatten) 펼치기
# 임의의 깊이로 중첩된 리스트를 입력받아, 모든 원소를 1차원 리스트로 반환하는 함수를 작성하세요.

def flatten(data) :
    result = []
    for i in data :
        if type(i) == list :
            result.extend(flatten(i))
        else :
            result.append(i)
    return result

a = [1, [2, 3, [4, 5, [6, 7]]], [8, 9]]

print(flatten(a))
```

```python
# 5. 문자열 내 단어 빈도수 세기
# 문자열을 입력받아, 각 단어가 몇 번 나오는지 세어 딕셔너리로 반환하는 함수를 작성하세요.

def word_freauency(sentence) :
    words = sentence.split()
    print(words)
    words_count ={}
    for word in words :
        if word in words_count :
            words_count[word] += 1
        else :
            words_count[word] = 1
    return words_count

sentence = "Hello world! This is a test sentence. Hello again!"

print(word_freauency(sentence))
```

```python
def longest_word (str1, str2) :
    intersection_word = {}
    for word in str1 :
        if word in str2 :
            intersection_word[word] = len(word)
        else :
            continue
    for k, v  in intersection_word.items() :
        if v == max(intersection_word.values()) :
            return k, v

str1 = 'apple', 'banana', 'grape', 'mango', 'orange'
str2 = 'grape', 'kiwi', 'mango', 'pear', 'orange'

print(longest_word(str1, str2))
```
