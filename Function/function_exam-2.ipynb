```python
# 1. 소수(prime number) 판별 함수
# 정수 n을 입력받아, n이 소수이면 True, 아니면 False를 반환하는 함수를 작성하세요.

def prime_number(n) :
    if n == 1 :
        return False
    for i in range(2, n) :
        if n % i == 0 :
            return False
    return True
        
print(prime_number(34))
print(prime_number(7))
```

```python
# 2. 중복 없는 리스트 반환
# 정수로 이루어진 리스트를 입력받아, 중복을 제거한 새로운 리스트를 반환하는 함수를 작성하세요(단, 원래의 순서를 유지할 것).

result = []

def remove_duplicates(data):
    for i in data :
        if i not in result :
            result.append(i)
    
    return result

a = [1, 3, 2, 5,34, 5, 6, 34, 3, 5, 43, 5,]

print(remove_duplicates(a))


  
```

```python
# 3. 단어별 첫 글자만 대문자로
# 문자열을 입력받아, 각 단어의 첫 글자만 대문자로 바꾼 문자열을 반환하는 함수를 작성하세요.

def capitalize_words(sentence):
    for i in sentence.split():
        sentence = sentence.replace(i, i.capitalize())
    return sentence

print(capitalize_words("hello world"))
```

```python
# 4. 피보나치 수열 n번째 항 구하기
# 정수 n을 입력받아, 피보나치 수열의 n번째 항을 반환하는 함수를 작성하세요. (단, n은 1부터 시작)

def fibonaci(n) :
    if n ==1 or n == 2:
        return 1
    else :
        return fibonaci(n - 1) + fibonaci(n - 2)
    
print(fibonaci(15))
```

```python
# memorize 를 사용하여 속도를 빠르게 하자면....

import numpy as n
memo = {}

def fibonaci(n) :
    if n in memo :
        return memo[n]
    if n == 1 or n ==2 :
        return 1
    else :
        memo[n] = fibonaci(n-1) + fibonaci(n-2)
        return memo[n]
    
print(fibonaci(100))
```

```python
# 더 간단하게 만들면 

memo = {1 : 1, 2 : 1}

def fibonaci(n) :
    if n in memo :
        return memo[n]
    else :
        memo[n] = fibonaci(n-1) + fibonaci(n-2)
        return memo[n]
    
print(fibonaci(100))
```

```python
# 5. 두 리스트의 교집합 반환
# 두 개의 정수 리스트를 입력받아, 두 리스트에 모두 존재하는 값만 모아 새로운 리스트로 반환하는 함수를 작성하세요(중복 없이).

result = []

def intersection(list1, list2) :
    for i in list1 :
        if i in list2 :
            result.append(i)
    return result

list1 = [1, 2, 3, 4, 5]
list2 = [2, 3, 4, 5, 6]

print(intersection(list1, list2))
```
