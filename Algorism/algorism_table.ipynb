```python
total = 6

def graph(untabled) :

    print(f'{untabled}에 진입')
    for tabled in range(2, min(10, untabled) + 1) :
        print(f'{tabled}명 앉음 {untabled - tabled} 사람 남음')
        graph(untabled - tabled)
        print(f'{untabled} 로 돌아옴옴')

graph(total)
```

```python
# 경우의 수를 세기 위해서 untabled 가 0 이 된 경우의 수를 카우트 한다.
# count = 0 을 만들고, 이것을 안으로 넣어서 계산하게 되므로 golbal counter 변수로 해준다.

total = 6
counter = 0

def graph(untabled) :

    if untabled == 0 :
        global counter
        counter += 1
    # print(f'{untabled}에 진입')
    for tabled in range(2, min(10, untabled) + 1) :
        # print(f'{tabled}명 앉음 {untabled - tabled} 사람 남음')
        graph(untabled - tabled)
        # print(f'{untabled} 로 돌아옴옴')
graph(total)
print(counter)
```

```python
# 5개의 경우에 수는 2명 앉히고 4명 앉힌 경우와
# 4명 앉히고 2 명 앉히는 경우의 2가지가 중복 된 숫자 이므로 이것을 없애기 위해서는
# 일단 앉힌 사람보다 적은 사람을 앉히는 경우를 없애면 된다.. 
# 앉힌 사람위 수도 계속 확인해야 하므로 함수의 매개변수로 넣는다... 

total = 6 

def graph(untabled, before_tabled) :
    counter = 0 
    for tabled in range(max(2, before_tabled), min(10, untabled) + 1 ) :
        graph(untabled - tabled, tabled)
        if untabled == 0 :
            counter += 1
        counter += graph(untabled - tabled, tabled)
    return counter

graph(total, 0)
print(counter)
```

```python
# 이것을 메모화 해서 조금 더 시간을 줄여보자..

import time

start_time = time.time()

total  = 100
memo = {}  # 일반적으로 key = 함수의 매개변수, value = 리턴 값으로 정한다. # key = (untabled, tabled) 로 정한다면 memo[(untabled, tabled)] 가 되겠다.

def graph(untabled, tabled_before) :
    if (untabled, tabled_before) in memo :
        return memo[(untabled, tabled_before)]

    result =  0 # value 값인 return 값은 항등원으로 초기화 해야 한다.
    if untabled == 0 :
        result += 1
    else :
        for tabled in range(max(2, tabled_before), min(10, untabled) + 1 ) :
            # print(tabled)
            result += graph(untabled - tabled, tabled)
    memo[(untabled, tabled_before)] = result
    return result

print(graph(total, 0))  # for 문을 도는 tabled 와 변수 tabled 가 같으므로 계산 값의 오차가 발생했다..

end_time = time.time()

print(f'Duration of process : {end_time - start_time}')

```

```python
total = 100
memo = {}

def graph(untabled, tabled):
    if (untabled, tabled) in memo:
        return memo[(untabled, tabled)]
    
    result = 0
    if untabled == 0:
        result += 1
    else:
        # 반복문 변수 이름을 다르게!
        for next_t in range(max(2, tabled), min(10, untabled) + 1):
            result += graph(untabled - next_t, next_t)
        memo[(untabled, tabled)] = result  # 올바른 키로 저장
    return result

print(graph(total, 0))  # 결과: 128724897
```

```python
total = 100

def graph(untabled, tabled):
    if untabled == 0:
        return 1  # 유효한 경우의 수 1 반환
    
    count = 0
    # 다음 숫자(next_t)는 현재 숫자(tabled) 이상, 10 이하
    for next_t in range(max(2, tabled), min(10, untabled) + 1):
        count += graph(untabled - next_t, next_t)  # 하위 문제 결과 누적
    return count

print(graph(total, 0))  # 출력: 128724897
```

```python
total = 100

def graph(untabled, tabled):
    count = 0

    # 다음 숫자는 현재 숫자(tabled) 이상, 10 이하
    start = max(2, tabled)
    end = min(10, untabled) if untabled > 0 else 0  # untabled=0 처리

    for next_t in range(start, end + 1):
        remaining = untabled - next_t

        if remaining == 0:
            count += 1  # next_t가 untabled와 정확히 일치하는 경우
        elif remaining > 0:
            count += graph(remaining, next_t)  # 재귀 호출

    # 초기 untabled=0인 경우 (예: graph(0, 0))
    if untabled == 0 and tabled == 0:
        count = 1

    return count

print(graph(total, 0))  # 출력: 128724897
```
