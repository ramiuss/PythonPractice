```python
# abs() 함수 설명
# abs() 함수는 입력된 숫자의 절댓값(absolute value)을 반환하는 함수입니다. 
# 절댓값이란 어떤 수가 0에서 떨어진 거리, 즉 부호(+, -)를 무시한 수의 크기를 의미합니다. 
# 그래서 양수는 그대로, 음수는 양수로 변환됩니다. 예를 들어, -5의 절댓값은 5이고, 3의 절댓값은 3입니다.

# 사용 목적: 숫자의 부호와 관계없이 크기만 필요할 때 사용합니다. 예를 들어, 손익 계산, 거리 계산, 데이터 분석 
# 등에서 널리 활용됩니다.
```

```python
# 정수
print(abs(-5))  # 출력: 5

# 실수
print(abs(-3.7))  # 출력: 3.7

# 복소수
print(abs(complex(3, -4)))  # 출력: 5.0 (피타고라스 정리: sqrt(3^2 + 4^2))
```

```python
diff = abs(10 - 15)  # 결과: 5

print(diff)
```

```python
numbers = [3, -1, -4, 1, 5, -9]
abs_numbers = [abs(x) for x in numbers]  # 결과: [3, 1, 4, 1, 5, 9]

print(abs_numbers)
```

```python
# aiter() 함수 설명
# aiter() 함수는 Python 3.10부터 도입된 내장 함수로, 비동기 이터러블(asynchronous iterable) 객체의 비동기 이터레이터(asynchronous iterator)를 반환합니다. 동기 반복에 사용되는 iter() 함수의 비동기 버전으로, 주로 async for 문에서 비동기적으로 데이터를 반복 처리할 때 활용됩니다.

# 시그니처: aiter(async_iterable)

# 매개변수: 비동기 이터러블 객체 (예: 비동기 제너레이터, aiter 메서드를 가진 클래스 등)

# 반환값: 비동기 이터레이터 객체 (anext() 메서드를 통해 값을 비동기적으로 가져올 수 있음)

# 특징
# 동기(iter)와 달리, 비동기 이터러블만 지원합니다.

# async for 문 또는 await anext()와 함께 사용해 비동기 반복을 구현할 수 있습니다.

# 비동기 I/O, 네트워크, 대용량 데이터 처리 등에서 유용하게 쓰입니다.
```

```python
async def agen():
    yield 1
    yield 2
    yield 3

ait = aiter(agen())

async for v in ait:
    print(v)
# 출력: 1 2 3
```

```python
class AsyncIterable:
    def __aiter__(self):
        self.i = 0
        return self
    async def __anext__(self):
        if self.i < 3:
            self.i += 1
            return self.i
        else:
            raise StopAsyncIteration

ait = aiter(AsyncIterable())

async for v in ait:
    print(v)
# 출력: 1 2 3
```

```python
async def agen():
    for i in range(3):
        yield i

ait = aiter(agen())
print(await anext(ait))  # 0
print(await anext(ait))  # 1
print(await anext(ait))  # 2
```

```python
import aiofiles

async def read_file():
    async with aiofiles.open('test.txt') as f:
        ait = aiter(f)
        async for line in ait:
            print(line.strip())
```

```python
# 1. Python의 all() 함수
# 설명:
# all()은 **iterable(반복 가능한 객체)**의 모든 요소가 참(True)인지 확인하는 내장 함수입니다.
# 모든 요소가 참이면 True, 하나라도 거짓(False)이면 False를 반환합니다.
```

```python
numbers = [1, 2, 3, 4]
print(all(numbers))  # True, 모든 요소가 참(0이 아님)

values = [True, False, True]
print(all(values))   # False, False가 하나 있음

empty = []
print(all(empty))    # True, 빈 리스트는 기본적으로 True
```

```python
# anext() 함수 설명
# anext() 함수는 Python 3.10부터 도입된 비동기 이터레이터용 내장 함수입니다.
# 동기 이터레이터에서 사용하는 next()와 유사하지만, 비동기 이터레이터(async iterator)나 
# 비동기 제너레이터(async generator)에서 값을 하나씩 받아올 때 사용합니다. 반드시 await와 함께 사용해야 하며, 
# 비동기 프로그래밍에서 순차적으로 값을 받아오거나, 수동으로 이터레이션을 제어할 때 유용합니다.
```

```python
import asyncio

async def my_gen():
    for i in range(3):
        yield i

async def main():
    agen = my_gen()
    print(await anext(agen))  # 0
    print(await anext(agen))  # 1
    print(await anext(agen))  # 2
    try:
        print(await anext(agen))  # StopAsyncIteration 발생
    except StopAsyncIteration:
        print('No more items!')

asyncio.run(main())
```

```python
import asyncio

async def my_gen():
    yield 1

async def main():
    agen = my_gen()
    print(await anext(agen, '끝!'))  # 1
    print(await anext(agen, '끝!'))  # '끝!'

asyncio.run(main())
```

```python
import asyncio

class Counter:
    def __init__(self, stop):
        self.cur = 0
        self.stop = stop
    def __aiter__(self):
        return self
    async def __anext__(self):
        if self.cur < self.stop:
            result = self.cur
            self.cur += 1
            return result
        else:
            raise StopAsyncIteration

async def main():
    counter = Counter(2)
    print(await anext(counter))        # 0
    print(await anext(counter))        # 1
    print(await anext(counter, 'END')) # 'END'

asyncio.run(main())
```

```python
# any() 함수 설명
# any() 함수는 파이썬의 내장 함수로, 인자로 받은 반복 가능한 객체(iterable) 안에 하나라도 참(True)인 요소가 있으면 True를 반환합니다. 
# 모든 요소가 거짓(False)이거나 iterable이 비어 있으면 False를 반환합니다. 논리적으로 or 연산자와 비슷하게 동작합니다.

# 참으로 간주되는 값: 0이 아닌 수, True, 비어있지 않은 문자열/리스트/튜플 등

# 거짓으로 간주되는 값: 0, False, None, 빈 문자열/리스트/튜플 등
```

```python
print(any([False, False, False]))      # False (모두 거짓)
print(any([False, True, False]))       # True (하나라도 참)
print(any([0, '', None]))              # False (모두 거짓)
print(any([0, 'hello', None]))         # True ('hello'는 참)
print(any([]))                         # False (빈 iterable)
```

```python
lst = [1, 2, 3, 4, 5]
if any(x >= 10 for x in lst):
    print("10 이상의 요소가 있습니다.")
else:
    print("10 이상의 요소가 없습니다.")
```

```python
print(any("hello"))        # True (문자열이 비어있지 않으므로)
print(any(""))             # False (빈 문자열)
print(any((False, True)))  # True (하나라도 참)
print(any((False, False))) # False (모두 거짓)
```

```python
# ascii() 함수 설명
# ascii() 함수는 파이썬의 내장 함수로, 주어진 객체의 문자열 표현을 ASCII 문자만 포함하는 형태로 반환합니다. 
# 즉, 문자열이나 객체에 비-ASCII(영문 알파벳, 숫자, 특수문자 이외의 문자) 문자가 포함되어 있을 경우, 
# 해당 문자를 유니코드 이스케이프 시퀀스(
# 예: \uXXXX, \UXXXXXXXX)로 변환해서 반환합니다. 이 함수는 객체의 내용을 안전하게 ASCII 문자로 표현하고 싶을 때, 
# 특히 디버깅이나 로그 기록, 파일 저장 등에 유용하게 사용됩니다.
```

```python
# 한글 문자열 처리
text = "파이썬"
print(ascii(text))  # 출력: '\ud30c\uc774\uc36c'

# 이모지 등 특수문자 포함
s = "Hello, world! 👋"
print(ascii(s))  # 출력: 'Hello, world! \U0001f44b'

# 리스트, 딕셔너리 등 다양한 객체에도 사용 가능
data = {'국어': 100, '영어': 90}
print(ascii(data))  # 출력: "{'\uad6d\uc5b4': 100, '\uc601\uc5b4': 90}"

# 영어만 있을 때는 그대로 반환
print(ascii('Python'))  # 출력: 'Python'
```

```python
non_ascii = "눈사람☃️"
print(repr(non_ascii))  # '눈사람☃️'
print(ascii(non_ascii)) # '\ub208\uc0b0\ub9bc\u2603\ufe0f'
```
