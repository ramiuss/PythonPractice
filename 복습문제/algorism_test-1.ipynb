```python
# 어떤 마을에 100개의 집이 일렬로 나란히 있습니다.
# 각 집에는 스위치가 하나씩 있고, 처음에는 모두 꺼져 있습니다(OFF).

# 1번부터 100번까지의 집 앞을 1번 학생부터 100번 학생까지 각각 한 번씩 지나갑니다.

# 1번 학생은 모든 집의 스위치를 누릅니다(1, 2, 3, ..., 100번 집).

# 2번 학생은 2의 배수 집(2, 4, 6, ..., 100번 집)의 스위치를 누릅니다.

# 3번 학생은 3의 배수 집(3, 6, 9, ..., 99번 집)의 스위치를 누릅니다.

# ...

# 100번 학생은 100번 집의 스위치만 누릅니다.

# 각 학생이 지나갈 때마다 해당 집의 스위치는 ON/OFF가 바뀝니다.

# 질문
# 마지막에 스위치가 켜져 있는 집의 번호는 모두 몇 번일까요?
```

```python
# 1부터 20까지의 숫자가 적힌 카드가 일렬로 놓여 있습니다.
# 처음에는 모든 카드가 뒷면(숫자가 안 보이는 면)으로 놓여 있습니다.

# 이번에는 1번부터 20번까지의 사람이 차례로 나와서 다음과 같이 행동합니다.

# 각 사람은 자신의 번호와 "서로소(공약수가 1)"인 카드만 뒤집습니다.

# 예를 들어, 6번 사람은 1, 5, 7, 11, 13, 17, 19번 카드만 뒤집습니다.
# (6과 서로소인 숫자만 뒤집음)

# 모든 사람이 다 지나간 후,
# 앞면(숫자가 보이는 면)이 보이는 카드의 번호는 무엇일까요?
```

```python
# 서로소 를 찾기 위해 math 라이브러리를 이용해서 gcd
import math

n = 20
result = {}
for card in range(1, n + 1) :
    count = 0
    for person in range( 1, n + 1) :
        if math.gcd(card, person) == 1 :
            count += 1
            result[card] = count

for k, v in result.items() :
    if v % 2 == 1 :
        print(k, end=', ')
```

```python
n = 20

def gcd(a, b) :
    while b != 0 :
        a, b = b, a % b
    return a

front_card = {}

for card in range(1, 20 + 1) :
    count = 0
    for person in range(1, n + 1) :
        if gcd(card, person) == 1 :
            count += 1
            front_card[card] = count

print(front_card)

for k, v in front_card.items() :
    if v % 2 == 1 :
        print(k, end=', ')
```

```python
import math

n = 20
front_cards = []

for card in range(1, n + 1):
    count = 0
    for person in range(1, n + 1):
        if math.gcd(card, person) == 1:  # 서로소 판별
            count += 1
    if count % 2 == 1:  # 홀수 번 뒤집힌 카드만 앞면
        front_cards.append(card)

print("앞면이 보이는 카드 번호:", front_cards)
```
