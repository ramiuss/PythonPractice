```python
# 1. 도서 카테고리별 총 페이지 수 계산

books = [
    {"title": "Python Basics", "category": "Programming", "pages": 300},
    {"title": "Data Science 101", "category": "Data", "pages": 450},
    {"title": "Web Development", "category": "Programming", "pages": 280},
    {"title": "ML Fundamentals", "category": "Data", "pages": 520}
]

category_pages = {}
for i in books :
    k = i['category']
    v = i['pages']
    if k not in category_pages :
        category_pages[k] = []
    category_pages[k].append(v)
print(category_pages)

sum_pages = {}
for k, v in category_pages.items() :
    sum_pages[k] = sum(v)
print(sum_pages)
```

```python
# 2. 지역별 최고 온도 찾기

weather_data = [
    {"city": "Seoul", "region": "Capital", "temp": 28},
    {"city": "Busan", "region": "South", "temp": 32},
    {"city": "Daegu", "region": "South", "temp": 34},
    {"city": "Incheon", "region": "Capital", "temp": 26}
]

region_temp = {}
for i in weather_data :
    k = i['region']
    v = i['temp']
    if k not in region_temp :
        region_temp[k] = []
    region_temp[k].append(v)
print(region_temp)

max_temp = {}
for k, v in region_temp.items() :
    max_temp[k] = max(v)
print(max_temp)
```

```python
# 3. 사용자 계정 상태별 카운트

users = [
    {"id": 1, "status": "active", "join_date": "2023-01"},
    {"id": 2, "status": "inactive", "join_date": "2023-02"},
    {"id": 3, "status": "active", "join_date": "2023-03"},
    {"id": 4, "status": "suspended", "join_date": "2023-01"}
]

user_status = {}
for i in users :
    k = i['id']
    v = i['status']
    # if k not in user_status :
    #     user_status[k] = []
    user_status[k]= v
print(user_status)

status_count = {}
for k, v in user_status.items() :
    count = 0
    if v not in status_count :
        status_count[v] = 1
    else : 
        status_count[v] += 1
print(status_count)
```

```python
# 4. 제조사별 제품 가격 평균

products = [
    {"model": "X1", "maker": "Samsung", "price": 120},
    {"model": "A5", "maker": "LG", "price": 95},
    {"model": "X2", "maker": "Samsung", "price": 150},
    {"model": "B2", "maker": "LG", "price": 110}
]

model_price = {}
for i in products :
    k = i['maker']
    v = i['price']
    if k not in model_price :
        model_price[k] = []
    model_price[k].append(v)
print(model_price)

average_price = {}
for k, v in model_price.items() :
    average_price[k] = sum(v) / len(v)
print(average_price)
```

```python
# 5. 연령대별 관심사 통계

survey = [
    {"age": 25, "interest": "sports"},
    {"age": 32, "interest": "music"},
    {"age": 25, "interest": "tech"},
    {"age": 40, "interest": "sports"}
]

age_interest = {}
for i in survey :
    k = i['age']
    v = i['interest']
    if k not in age_interest :
        age_interest[k] = []
    age_interest[k].append(v)
print(age_interest)
```

```python
# 1. 홀짝 연속 그룹핑

nums = [2, 4, 6, 3, 5, 8, 7, 9]
# 출력 예시: [[2,4,6], [3,5], [8], [7,9]]

result = []
inner_result = [nums[0]]

for i in nums[1:] :
    if i % 2 == inner_result[-1] % 2 :
        inner_result.append(i)
    else :
        result.append(inner_result) 
        inner_result = [i]
result.append(inner_result)
print(result)
    
```

```python
# 2. 부호 변화 그룹핑

sequence = [1, 2, -3, -4, 5, -6, -7]
# 출력 예시: [[1,2], [-3,-4], [5], [-6,-7]]

result = []
inner_result = [sequence[0]]

for i in sequence[1:] :
    if i * inner_result[-1] > 0 :
        inner_result.append(i)
    else :
        result.append(inner_result)
        inner_result = [i]
result.append(inner_result)
print(result)
```

```python
# # 3. 모음 연속 그룹핑

chars = ['a', 'a', 'i', 'b', 'e', 'e', 'c', 'u']
# 출력 예시: [['a','a','i'], ['b'], ['e','e'], ['c'], ['u']]

result = []
inner_result = [chars[0]]
for i in chars[1:] :
    if (i in 'a,e,i,o,u') == (inner_result[0] in 'a,e,i,o,u') :
        inner_result.append(i)
    else : 
        result.append(inner_result)
        inner_result = [i]
result.append(inner_result)
print(result)
```

```python
# 4. 이진수 패턴 그룹핑

binary = [0,0,1,1,0,1,0,0,0]
# 출력 예시: [[0,0], [1,1], [0], [1], [0,0,0]]

result = []
inner_result = [binary[0]]
for i in binary[1:] :
    if i == inner_result[-1] :
        inner_result.append(i)
    else :
        result.append(inner_result)
        inner_result = [i]
result.append(inner_result)
print(result)
    

```

```python
# 5. 단계적 증가 그룹핑

numbers = [10,11,12,15,16,20,21]
# 출력 예시: [[10,11,12], [15,16], [20,21]]

result = []
inner_result = [numbers[0]]
for i in numbers[1:] :
    if i -1 == inner_result[-1] :
        inner_result.append(i)
    else :
        result.append(inner_result)
        inner_result = [i]
result.append(inner_result)
print(result)


```
