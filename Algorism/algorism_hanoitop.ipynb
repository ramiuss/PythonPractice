```python
# 하노이 탑 문제... 

num = 3

def hanoi_top(num, start, desti, inter) :
    if num == 1 :
        print(f'{start} -> {desti}')
    else :
        hanoi_top(num - 1, start, inter, desti)
        print(f'{start} -> {desti}')
        hanoi_top(num - 1, inter, desti, start)

hanoi_top(num, 'A', 'B', 'C')
```

```python
num = 3
count = 0

def hanoi_top(num, start, desti, inter) :
    global count
    if num == 1 :
        count += 1
        print(f'{start} -> {desti}. {count}')
  
    else :
        hanoi_top(num - 1, start, inter, desti)
        count += 1
        print(f'{start} -> {desti}, {count}')

        hanoi_top(num - 1, inter, desti, start)
    return count

hanoi_top(num, 'A', 'B', 'C')
print(count)
```
