```python
# 1. 학생 점수 평균 구하기
# 아래와 같은 딕셔너리가 있습니다.
# 각 학생의 평균 점수를 구해서 리스트로 출력하세요.

scores = {
    "Tom": [80, 75, 90],
    "Jane": [85, 95, 100],
    "Mike": [70, 60, 80]
}
# 출력 예시: [81.67, 93.33, 70.0]

def average_scores(data) :
    result = []
    for k, v in data.items() :
        result.append((k , round(sum(v)/len(v), 2)))
    return result
print(average_scores(scores))
print()

scores_values = scores.values()
print((scores_values))   #이렇게 하면 dict_values 값을 표시하게 되므로 리스트로 전환해야 한다
print(list(scores_values))

# 따라서 처름에 리스트로 만들어 저장을 한다면..
scores_list = list(scores.values())
print(scores_list)

for i in scores_list :
    print(round(sum(i)/len(i), 2 ), end=', ')
print()


print(list(map(lambda x: round(sum(x)/len(x), 2), (i for i in list(scores.values())))))

print(list(map(lambda x : round(sum(i)/len(i), 2) for i in list(scores.values()))))


    

```

```python
# 2. 과일 개수 세기
# 다음과 같은 리스트가 있습니다.
# 각 과일이 몇 번 나왔는지 딕셔너리로 세어 출력하세요.

fruits = ["apple", "banana", "apple", "orange", "banana", "apple"]
# 출력 예시: {'apple': 3, 'banana': 2, 'orange': 1}

def count_word(data) :
    result = {}
    count = 0
    for i in data :
        if i not in result :
            count = 1
            result[i] = count
        else :
            count += 1
            result[i] = count
    return result
print(count_word(fruits))

def count_words(data) :
    result = {}
    for i in data :
        if i not in result :
            result[i] = 1
        else :
            result[i] += 1
    return result
print(count_words(fruits))

```

```python
# 3. 학생 이름 오름차순 정렬
# 아래 딕셔너리에서 학생 이름(키)을 오름차순으로 정렬하여
# 이름과 점수를 한 줄씩 출력하세요.

scores = {
    "David": 88,
    "Alice": 92,
    "Bob": 79
}
# 출력 예시:
# Alice 92
# Bob 79
# David 88

def sort_student_name(data) :
    for i in sorted(data) :
        print(i, data[i])
sort_student_name(scores)

def sort_student_names(data) :
    result = {}
    for i in sorted(data) :
        result[i] = data[i]
    return result
print(sort_student_names(scores))

```

```python
# 4. 중복 없는 리스트 만들기
# 아래 리스트에서 중복을 제거한 새로운 리스트를 만들어 출력하세요.

numbers = [1, 2, 2, 3, 4, 4, 5, 1, 6]
# 출력 예시: [1, 2, 3, 4, 5, 6]

def simple_numbers(data) :
    result = []
    for i in data :
        if i not in result :
            result.append(i)
    return result
print(simple_numbers(numbers))


```

```python
# 5. 점수 80점 이상 학생만 딕셔너리로 만들기
# 아래 딕셔너리에서 점수가 80점 이상인 학생만 뽑아
# 새로운 딕셔너리로 만들어 출력하세요.

scores = {
    "Anna": 78,
    "Brian": 82,
    "Cathy": 91,
    "Derek": 65
}
# 출력 예시: {'Brian': 82, 'Cathy': 91}

def superior_students(data) :
    result = {}
    for k, v in data.items() :
        if v >= 80 :
            result[k] = v
    return result 
print(superior_students(scores))
```

```python
# 1. 학생별 최고 점수 찾기
# 아래 딕셔너리에서 각 학생의 최고 점수만 모아 새로운 딕셔너리를 만들어 출력하세요.

scores = {
    "Jin": [85, 90, 78],
    "Mina": [88, 92, 95],
    "Lee": [70, 80, 65]
}
# 출력 예시: {'Jin': 90, 'Mina': 95, 'Lee': 80}

def bset_score(data) :
    result = {}
    for k, v in data.items() :
        for i in v :
            if i == max(v) :
                result[k] = i
    return result
print(bset_score(scores))
```

```python
# 2. 리스트에서 짝수만 골라내기
# 아래 리스트에서 짝수만 뽑아 새로운 리스트로 만드세요.

numbers = [3, 6, 8, 1, 5, 10, 12, 7]
# 출력 예시: [6, 8, 10, 12]

def find_evev(data) :
    result = []
    for i in data :
        if i % 2 == 0 :
            result.append(i)
    return result
print(find_evev(numbers))
```

```python
# 3. 딕셔너리 값의 합 구하기
# 아래 딕셔너리의 값(점수)들의 총합을 구하세요.

scores = {
    "Kim": 75,
    "Park": 88,
    "Choi": 92
}
# 출력 예시: 255

def sum_scores(data) :
    result = 0
    for k, v in data.items() :
        result += v 
    return result
print(sum_scores(scores))
```

```python
# 4. 리스트에서 가장 많이 나온 값 찾기
# 아래 리스트에서 가장 많이 나온 값을 찾아 출력하세요.

nums = [1, 3, 3, 2, 5, 3, 2, 4, 2, 2]
# 출력 예시: 2

def many_number(data) :
    result = {}
    count = 0
    for i in data :
        if i not in result :
            result[i] = 1
        else :
            result[i] += 1
    for k, v in result.items() :
        if v == max(result.values()) :
            return k
print(many_number(nums))

```

```python
# 5. 두 리스트의 합집합 만들기
# 아래 두 리스트의 모든 원소를 합쳐서 중복 없이 하나의 리스트로 만드세요.

a = [1, 2, 3, 4]
b = [3, 4, 5, 6]
# 출력 예시: [1, 2, 3, 4, 5, 6]

def combine_ab(data1, data2) :
    result = []
    for i in data1 :
        if i not in data2 :
            result.append(i)
    for j in data2 :
        if j not in result :
            result.append(j)
    return result
print(combine_ab(a, b))
```
