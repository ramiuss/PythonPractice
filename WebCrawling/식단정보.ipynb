```python
import requests
from bs4 import BeautifulSoup
```

```python
url = "https://school.cbe.go.kr/cbs-h/M01050705/list"
res = requests.get(url)
print(res.status_code)
print(res.headers)
```

```python
html = res.text
html
```

```python
soup = BeautifulSoup(html, 'html.parser')
soup
```

```python
base_url ="https://school.cbe.go.kr/cbs-h/M01050705/list?ymd="
date = '20250922'

res = requests.get(base_url + date)
print(res.status_code)
print(res.headers)
```

```python
html = res.text
soup = BeautifulSoup(html, 'html.parser')
print(soup)
```

```python
meals = soup.find_all( class_='tch-lnc-wrap')
```

```python
for i in meals :
    print(i.text)
```
