```python
# callable() 함수란?
# **callable()**은 파이썬에서 객체가 호출 가능한지—즉, 함수처럼 소괄호 ()를 붙여 실행할 수 있는지—를 판별하는 내장 함수다. 호출이 가능하면 True, 아니면 False를 반환한다.

# 호출 가능한 객체(callable object)는 다음과 같은 것들이 있다:

# 함수(내장 함수, 사용자 정의 함수, 람다 등)

# 메서드

# 클래스(클래스를 호출하면 인스턴스가 생성됨)

# call 메서드를 구현한 클래스의 인스턴스
```

```python
# 1. 함수는 호출 가능하다
def foo():
    pass

print(callable(foo))  # True

# 2. 내장 함수도 호출 가능하다
print(callable(len))  # True

# 3. 리스트와 같은 일반 객체는 호출 불가
print(callable([1, 2, 3]))  # False

# 4. 클래스는 호출 가능하다 (인스턴스 생성 가능)
class MyClass:
    pass

print(callable(MyClass))  # True

# 5. __call__ 메서드를 가진 객체는 호출 가능하다
class CallableDemo:
    def __call__(self):
        print("I'm called!")

cd = CallableDemo()
print(callable(cd))  # True

# 6. __call__ 메서드가 없는 일반 객체는 호출 불가
class NotCallable:
    pass

nc = NotCallable()
print(callable(nc))  # False

# 7. lambda(람다) 함수도 callable
f = lambda x: x * 2
print(callable(f))  # True
```

```python
class Person:
    def __init__(self, name):
        self.name = name
    def __call__(self, major):
        print(f"{self.name}의 전공은 {major}입니다.")

p = Person("홍길동")
p("컴퓨터공학")  # 홍길동의 전공은 컴퓨터공학입니다.
print(callable(p))  # True
```

```python
# chr() 함수란?
# chr() 함수는 파이썬의 내장 함수로, *정수(유니코드 코드 포인트)*를 해당하는 문자로 변환해 반환한다. 즉, 숫자를 문자로 바꿀 때 사용한다.

# 입력값은 0부터 1,114,111(0x10FFFF)까지의 정수여야 하며, 이 범위를 벗어나면 ValueError가 발생한다.

# 반대로, 문자를 정수(유니코드 코드 포인트)로 바꾸고 싶을 때는 ord() 함수를 사용한다.
```

```python
print(chr(65))    # A
print(chr(97))    # a
print(chr(38))    # &
print(chr(1200))  # Ұ
```

```python
print(chr(-1))        # 범위를 벗어나면 ValueError 발생
print(chr(1114112))   # 범위를 벗어나면 ValueError 발생
```

```python
# 대문자 알파벳 모두 생성
letters = [chr(i) for i in range(65, 91)]
print(letters)  # ['A', 'B', ..., 'Z']
```

```python
def ascii_to_char(ascii_list):
    return ''.join([chr(i) for i in ascii_list])

print(ascii_to_char([65, 66, 67]))  # "ABC"
```

```python
def char_shift(char, shift):
    return chr(ord(char) + shift)

print(char_shift('A', 2))  # "C"

```

```python
# classmethod()란?
# **classmethod()**는 파이썬에서 메서드를 클래스 메서드로 만들어주는 내장 함수이자, 데코레이터(@classmethod)로도 많이 사용된다. 클래스 메서드는 인스턴스가 아니라 클래스 자체를 첫 번째 인자(cls)로 받아, 클래스 변수나 클래스 상태를 조작할 수 있다.

# 특징
# 첫 번째 인자로 클래스를 받으며, 관례적으로 cls라고 쓴다.

# 인스턴스 없이도 클래스 이름으로 직접 호출할 수 있다.

# 클래스 변수(클래스 전체가 공유하는 변수)에 접근하거나, 클래스 전체에 영향을 주는 작업에 주로 사용된다.

# 팩토리 메서드(다양한 방식으로 인스턴스를 생성하는 함수) 구현에도 자주 쓰인다.
```

```python
class MyClass:
    class_variable = 0

    @classmethod
    def increase_class_variable(cls):
        cls.class_variable += 1

MyClass.increase_class_variable()
print(MyClass.class_variable)  # 출력: 1
```

```python
class MyClass:
    class_variable = 0

    def increase_class_variable(cls):
        cls.class_variable += 1

    increase_class_variable = classmethod(increase_class_variable)

MyClass.increase_class_variable()
print(MyClass.class_variable)  # 출력: 1
```

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    @classmethod
    def square(cls, side):
        return cls(side, side)

r = Rectangle.square(5)
print(r.width, r.height)  # 출력: 5 5
```
