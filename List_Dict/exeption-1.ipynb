```python
# 사용자로 부터 화일 이름을 받고.
# 그 파일의 내용을 int(0 함수로 정수로 변환한 뒤 출력하는 프로그램

def read_file(name_file) :
    try :
        file = open(name_file, 'r', encoding = 'utf-8')
    except :
        print('There is no that name file')
    else : 
        item = file.read()
        try :
            number = int(item)
            return number
        except :
            print('There no number')
        finally :
            file.close()
  

file_name = input('화일 이름을 넣어 주세요 : ')
print(read_file(file_name))
```

```python
# 4. 함수에서 예외 처리

# 아래 요구사항을 만족하는 함수를 작성하세요.
# 함수 이름은 divide(a, b)입니다.
# a를 b로 나누어 반환합니다.
# b가 0이면 "0으로 나눌 수 없습니다."를 반환하세요.
# 예외가 발생하지 않으면 "나눗셈 성공"을 출력하세요.
# 항상 "함수 종료"를 출력 합니다.

def divide(a, b) :
    print(a / b )

a, b = map(int, input('숫자 2개를 입력해 주세요 : ').split())

try :
    divide(a, b)
except ZeroDivisionError :
    print('0으로  나눌 수 없습니다.. ')
else :
    print('나눗셈 성공..')
finally :
    print('프로그램 종료.')

```

```python
# 5. 여러 예외 처리

# 아래 요구사항을 만족하는 코드를 작성하세요.
# 사용자에게 두 개의 정수를 입력받아 나눗셈을 합니다.
# 입력값이 정수가 아니면 "정수를 입력하세요."를 출력하세요.
# 두 번째 숫자가 0이면 "0으로 나눌 수 없습니다."를 출력하세요.
# 예외가 발생하지 않으면 결과를 출력하고 "연산 성공"을 출력하세요.
# 항상 "프로그램 종료"를 출력하세요.
# 필요하다면 각 문제에 대해 예시 입력/출력도 제공해 드릴 수 있습니다!

try:
    # 두 개의 정수를 한 줄에 입력받아 처리
    num1, num2 = map(int, input('두 개의 정수를 입력하세요 (예: 10 2): ').split())
    result = num1 / num2
except ValueError:
    print('정수를 입력하세요.')
except ZeroDivisionError:
    print('0으로 나눌 수 없습니다.')
else:
    print('결과:', result)
    print('연산 성공')
finally:
    print('프로그램 종료')
```
