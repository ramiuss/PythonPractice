```python
# 문자열의 성질(숫자, 알파벳, 공백 등) 확인: isdigit(), isalpha(), isalnum(), isspace(), isnumeric(), isdecimal(), islower(), isupper(), startswith(), endswith()

# 컬렉션에서 값 존재 확인: in, any(), all()

# 부분 문자열 포함 여부: in, any(sub in item for item in lst)
```

```python
# isdigit()

# 문자열이 모두 숫자(0~9)로만 이루어져 있는지 확인합니다.

txt = "50800"
print(txt.isdigit())  # True
```

```python
# isdecimal()

# 문자열이 모두 10진수 숫자로만 이루어져 있는지 확인합니다.
# digit()보다 더 엄격하게 10진수만 인정합니다.

txt = "12345"
print(txt.isdecimal())  # True
```

```python
# isnumeric()

# 문자열이 숫자(정수, 분수, 로마 숫자, 지수 등)로만 이루어져 있는지 확인합니다.

txt = "Ⅻ"  # 로마 숫자 12
print(txt.isnumeric())  # True
```

```python
# isalpha()

# 문자열이 모두 알파벳 문자로만 이루어져 있는지 확인합니다.

txt = "Hello"
print(txt.isalpha())  # True
```

```python
# isalnum()

# 문자열이 모두 알파벳 또는 숫자로만 이루어져 있는지 확인합니다.

txt = "Hello123"
print(txt.isalnum())  # True
```

```python
# isspace()

# 문자열이 모두 공백 문자(스페이스, 탭, 줄바꿈 등)로만 이루어져 있는지 확인합니다.

txt = "   "
print(txt.isspace())  # True
```

```python
# islower(), isupper()

# 각각 문자열이 모두 소문자/대문자인지 확인합니다.
```

```python
# startswith(prefix), endswith(suffix)

# 문자열이 특정 접두사/접미사로 시작/끝나는지 확인합니다.

txt = "python"
print(txt.startswith("py"))  # True
print(txt.endswith("on"))    # True
```

```python
# in 연산자

# 부분 문자열의 포함 여부를 확인합니다.

"py" in "python"  # True

```

```python
# in 연산자

# 리스트, 튜플, 딕셔너리 등에서 값의 존재 여부를 확인합니다.

3 in [1, 2, 3]  # True
"a" in {"a":1, "b":2}  # True (딕셔너리는 키 기준)
```

```python
# any(), all()

# 반복 가능한 객체에서 하나라도 True인지, 모두 True인지 확인합니다.

any([0, 1, 2])  # True
all([1, 2, 3])  # True
```

```python
# 리스트 내에 특정 값이 존재하는지 확인하는 함수:

def isValInLst(val, lst):
    return val in lst
```

```python
# 부분 문자열이 리스트 내 문자열의 일부로 존재하는지 확인:

def substring_in_list(sub, lst):
    return any(sub in item for item in lst)
```
