```python
scores = {
    "Alice": [85, 90, 78],
    "Bob": [90, 88, 95],
    "Charlie": [70, 80, 65]
}
# 결과: {'Alice': 84.33, 'Bob': 91.0, 'Charlie': 71.67}, ('Bob', 91.0)

result = {}

def average_score(scores):
    for k, v in scores.items() :
        result[k] = sum(v) / len(v)
    return result

def best_student(result) :
    for k, v in result.items() :
        if v == max(result.values()) :
            return k, v

print(average_score(scores))
print(best_student(result))
```

```python
scores = {
    "Alice": [85, 90, 78],
    "Bob": [90, 88, 95],
    "Charlie": [70, 80, 65]
}

def average_score(data) :
    result = {}
    for k, v in data.items() :
        result[k] = round(sum(v) /len(v), 2)
    return result
result = average_score(scores)
print(result)

def best_student(data) :
    for k, v in data.items() :
        if v == max(data.values()) :
            return k, v
print(best_student(result))
```

```python
words = ["strawberry", "kiwi", "apple", "banana"]
# 여기에 코드를 작성하세요.

def long_word(data) :
    length = 0
    longest_word = []
    for i in data :
        if len(i) > length :
            length = len(i)
            longest_word.append(i)
        else :
            continue
    return longest_word

print(long_word(words))

print(sorted(words, key = lambda x : len(x)))
```

```python
# 1. 세제곱 리스트 만들기
# 정수 리스트 numbers가 주어졌을 때, map과 lambda를 이용해 모든 요소를 세제곱(3제곱)한 결과 리스트를 만드세요.

numbers = [1, 2, 3, 4, 5]

result = []
for i in numbers :
    result.append(i**3)
print(result)

def triple_power(data) :
    result = []
    for i in data :
        result.append(i**3)
    return result
print(triple_power(numbers))

print(list(map(lambda x : x**3 , numbers)))
```

```python
# 정수 리스트 numbers에서 filter와 lambda를 이용해 음수만 추출하는 코드를 작성하세요.

numbers = [-10, 3, -2, 7, -5, 0, 8]

result = []
for i in numbers :
    if i < 0 :
        result.append(i)
print(result)

def negative_number(data) :
    result = []
    for i in data :
        if i < 0 :
            result.append(i)
    return result
print(negative_number(numbers))

print(list(filter(lambda x : x < 0 , numbers)))
```

```python
# 3. 문자열 길이 4 이상만 대문자로 변환
# 문자열 리스트 words에서 filter로 길이가 4 이상인 단어만 걸러내고, map과 lambda로 대문자로 변환된 결과 리스트를 만드세요.

words = ['sun', 'moon', 'star', 'planet', 'sky']

result = []
for i in words :
    if len(i) >= 4 :
        result.append(i.upper())
print(result)

print(list(map(lambda x : x.upper(), filter(lambda x : len(x) >= 4, words))))
```

```python
# 4. 두 리스트의 동일 인덱스 요소 곱하기
# 두 정수 리스트 a, b가 있을 때, map과 lambda를 이용해 같은 인덱스의 요소끼리 곱한 결과 리스트를 만드세요

a = [1, 2, 3]
b = [4, 5, 6]

result = []

for i in range(len(a)) :
    result.append(a[i] * b[i])
print(result)

def list_power(data1, data2) :
    result = []
    for i in range(len(data1)) :
        result.append(data1[i] * data2[i])
    return result
print(list_power(a, b))

print(list(map(lambda x, y : x * y , a, b)))
```

```python
# 5. 홀수의 제곱만 추출
# 정수 리스트 numbers에서 filter와 lambda로 홀수만 추출하고, map과 lambda로 그 제곱값만 리스트로 만드세요.

numbers = [1, 2, 3, 4, 5, 6, 7]

result = []
for i in numbers :
    if i % 2 ==1 :
        result.append(i **2)
print(result)

def odd_power(data) :
    result = []
    for i in data :
        if i % 2 == 1:
            result.append(i ** 2)
    return result
print(odd_power(numbers))

print(list(map(lambda x : x **2 , filter(lambda x : x % 2 ==1, numbers))))
```

```python
# 1. 문자열 리스트에서 숫자 포함 단어만 소문자로 변환
# 문자열 리스트 words에서, 숫자가 하나라도 포함된 단어만 filter와 lambda로 추출하고, map과 lambda로 소문자로 변환한 리스트를 만드세요.

words = ['Hello1', 'world', 'Python3', 'lambda', 'Map2Filter']
# 결과: ['hello1', 'python3', 'map2filter']

# print(words[0][0])
result = []
for i in words :
    if any(j.isdigit() for j in i) :
        result.append(i.lower())
print(result)



```

```python
# 2. 튜플 리스트에서 두 값의 합이 10 이상인 튜플만 추출
# 아래와 같은 튜플 리스트 data에서, 두 값의 합이 10 이상인 튜플만 filter와 lambda로 추출하세요.

data = [(3, 7), (2, 4), (6, 5), (10, 1), (4, 4)]
# 결과: [(3, 7), (6, 5), (10, 1)]

result = []
for i in data :
    if i[0] + i[1] >= 10 :
        result.append(i)
print(result)

def add_tuple(data) :
    result = []
    for i in data :
        if i[0] + i[1] >= 10 :
            result.append(i)
    return result
print(add_tuple(data))

print(list(filter(lambda x :  x[0] + x[1] >= 10, data)))

```

```python
# 3. map, filter로 짝수의 제곱만 추출
# 정수 리스트 numbers에서, filter와 lambda로 짝수만 추출하고, map과 lambda로 제곱값만 리스트로 만드세요.

numbers = [1, 2, 3, 4, 5, 6, 7, 8]
# 결과: [4, 16, 36, 64]

result = []
for i in numbers :
    if i % 2 == 0:
        result.append(i **2)
print(result)  

def even_power(data) :
    result = []
    for i in data :
        if i % 2 == 0 :
            result.append(i ** 2) 
    return result
print(even_power(numbers))

print(list(map(lambda x : x**2 , filter(lambda x : x % 2 ==0, numbers))))
```

```python
# 4. 문자열 리스트에서 모음(a, e, i, o, u)으로 시작하는 단어만 대문자로 변환
# 문자열 리스트 words에서, filter와 lambda로 모음으로 시작하는 단어만 추출하고, map과 lambda로 대문자로 변환하세요.

words = ['apple', 'banana', 'orange', 'umbrella', 'grape', 'egg']
# 결과: ['APPLE', 'ORANGE', 'UMBRELLA', 'EGG']

print(words[0][0])

result = []
for i in words :
    if i[0] in ['a', 'e', 'i', 'o', 'u'] :
        result.append(i.upper())
print(result)


print(list(map(lambda x : x.upper(), filter(lambda x : x[0] in ['a', 'e', 'i', 'o', 'u'] , words))))
```

```python
# 5. 두 리스트에서 같은 인덱스의 값이 모두 홀수인 경우만 (a, b) 튜플로 묶기
# 정수 리스트 a, b가 있습니다. 같은 인덱스의 값이 모두 홀수인 경우만 (a, b) 형태의 튜플로 묶어 리스트로 만드세요. (lambda, map, filter 활용)

a = [1, 2, 3, 5, 8]
b = [7, 4, 9, 6, 11]
# 결과: [(1, 7), (3, 9)]

result = []
for i in range(len(a)) :
    if a[i] % 2 == 1 and b[i] % 2 == 1 :
        result.append((a[i], b[i]))
print(result)

print(list(map(lambda x : x,  filter(lambda x : x[0] % 2 == 1 and x[1] % 2 == 1, zip(a, b)))))
```
