```python
# 2. 다중 조건 필터링
# 1~100 사이 숫자에서 3과 5의 공배수를 제외한 3의 배수 또는 5의 배수만 추출한 후,
# 해당 숫자들의 1의 자리 수를 제곱한 리스트를 생성하세요.

# 예시 출력: [9, 25, 36, 9, 49, 81, 16, 64, ...]
```

```python
def single_digit(num1, num2) :
    result = []
    for i in range(num1, num2 + 1) :
        if (i % 3 == 0 or i % 5 == 0) and i % 15 != 0 :
            result.append((i % 10 ) ** 2)
    return result
print(single_digit(1, 100))

print(list(map(lambda x : (x % 10) ** 2 , filter(lambda x : (x % 3 == 0 or x % 5 == 0) and x % 15 != 0 , range( 1, 101)) )))
```

```python
# 3. 예외 처리 포함 변환
# 문자열 리스트에서 숫자로 변환 가능한 요소만 골라내어 3제곱한 값으로 매핑하되,
# 변환 실패 시 해당 요소를 제외시키는 코드를 작성하세요.

input_data = ['12', '3.14', 'abc', '45', '6.28j']
# 예시 출력: [1728, 31.003952, 91125]
```

```python
def make_number(data) :
    try :
        float(data) 
        return True
    except :
        return False
    
result = list(filter(make_number, input_data))
print(result)

print(list(map(lambda x : float(x) ** 3 , result)))
```

```python
# 4. 튜플 데이터 가공
# 다음 튜플 리스트에서 각 튜플 요소의 합이 짝수인 경우만 필터링하고,
# 남은 튜플들의 (첫 번째 요소 × 두 번째 요소) 결과를 구하세요.

data = [(3, 5), (2, 4), (7, 1), (6, 3), (8, 2)]
# 예시 출력: [15, 8, 6, 16]

def even_tuple(data) :
    result = []
    for i in data :
        if (i[0] + i[1]) % 2 == 0 :
            result.append(i[0] * i[1])
    return result
print(even_tuple(data))

print(list(map(lambda x : x[0] * x[1] , filter(lambda x : (x[0] + x[1]) % 2 == 0, data))))
```

```python
# 5. 계승수 자릿수 합 계산
# 1~10까지 숫자들의 계승값(factorial)을 구한 후,
# 각 결과값의 자릿수 합이 10보다 큰 숫자만 필터링하고
# 해당 계승값의 제곱근을 정수로 변환한 리스트를 생성하세요.

# 예시 출력: [2, 6, 24, 120, 720, 5040, 40320]
```

```python
def factorial(num) :
    factorial = []
    temp = 1
    for i in range(1, num + 1) :
        temp *= i
        factorial.append(temp)
    return factorial
result_1 = factorial(10)
print(result_1)
print()


a = 362880 
temp = 0
for i in str(a) :
    temp += int(i)
print(temp)
    

def str_number(data) :
    result = []
    for i in data :
        temp = 0
        for j in str(i) :
            temp += int(j) 
        result.append(temp)
    return result
result_2 = str_number(result_1)

print(result_2)
print()

def find_number(data) :
    result = []
    for i in data :
        temp = 0
        for j in str(i) :
            temp += int(j)
        if temp >= 10 :
            result.append(int(i ** 0.5))
    return result
print(find_number(result_1))


```
