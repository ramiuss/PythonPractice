```python
# 문제 1. 문자열 압축 (난이도: 중)
# 주어진 문자열에서 연속으로 나타나는 문자를 압축해서 표현하세요.
# 예를 들어, 문자열이 "aaabbccccdaa"라면 "a3b2c4d1a2"로 압축됩니다.

# 입력 예시
# "aaabbccccdaa"

# 출력 예시
# "a3b2c4d1a2"
```

```python
a ="aaabbccccdaa"

current_chr = a[0]
count = 1
result = []
for i in a[1::] :
    if i == current_chr :
        count += 1
    else :
        result.append(current_chr + str(count))
        current_chr = i
        count = 1
result.append(current_chr + str(count))
print(result)
print(f'정답은 {''.join(result)}')
```

```python
a ="aaabbccccdaafffffff"

def compress_string(data) :
    current_chr = data[0]
    count = 1
    result = []
    for i in data[1::] :
        if i == current_chr :
            count += 1
        else :
            result.append(current_chr + str(count))
            current_chr = i
            count = 1

    result.append(current_chr + str(count))

    return ''.join(result)


print(compress_string(a))
```

```python
# 문제 2. 두 리스트의 교집합 (난이도: 하)
# 두 개의 정수 리스트가 주어졌을 때, 두 리스트에 모두 존재하는 원소만을 담은 리스트를 반환하세요. 
# 단, 결과는 중복 없이 오름차순으로 정렬되어야 합니다.

# 입력 예시
# list1 = [1, 2, 3, 4, 5, 6, 7]
# list2 = [4, 5, 6, 7, 8, 9, 10]
```

```python
list1 = [1, 2, 3, 4, 5, 6, 7]
list2 = [4, 5, 6, 7, 8, 9, 10]

set_list1 = set(list1)
set_list2 = set(list2)
print(set_list1, type(set_list1))
print(set_list2, type(set_list2))

# 두개의 리스트를 합하면
print(set_list1 | set_list2)

# 두개의 리스트에 고통으로 들어가 있는 요소
print(set_list1 & set_list2)

# 차집합을 구하는 것은 
print(set_list1 - set_list2)
```

```python
# 1부터 50까지의 숫자가 적힌 카드가 각각 한 장씩 일렬로 놓여 있습니다.
# 처음에는 모든 카드가 뒷면(숫자가 안 보이는 면)으로 놓여 있습니다.

# 1번부터 50번까지의 사람이 차례로 나와서 다음과 같이 행동합니다.

# 1번 사람은 모든 카드(1, 2, ..., 50)를 뒤집습니다.

# 2번 사람은 2의 배수 카드(2, 4, 6, ..., 50)만 뒤집습니다.

# 3번 사람은 3의 배수 카드(3, 6, 9, ..., 48)만 뒤집습니다.

# ...

# 50번 사람은 50번 카드만 뒤집습니다.

# 모든 사람이 다 지나간 후,
# 앞면(숫자가 보이는 면)이 보이는 카드의 번호는 무엇일까요?
```

```python
card = 50
person = 50

result = {}
for i in range(1, card + 1) :
    count = 0
    for j in range(1, i + 1) :
        if i % j == 0 :
            count += 1
            result[i] = count 
print(result)

#숫자가 보인다는 것은 홀수 번 넘겼다는 말이 되므로..
answer = []
for k, v in result.items() :
    if v % 2 == 1:
        answer.append(k)

print(f'정답은 {answer}')
```

```python
def uncover_card(n) :
    count = 0
    for i in range(1, n + 1) :
        if n % i == 0 :
            count += 1
    return count

print(uncover_card(10))

n = 50
answer = []
for i in range(1, n + 1) :
    if uncover_card(i) % 2 == 1 :
        answer.append(i)

print(f'정답은 {answer}')
```
