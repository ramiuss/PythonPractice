```python
driver = webdriver.Chrome()
url = "https://google.com"
driver.get(url)
```

```python
action = ActionChains(driver)
```

```python
input_box = driver.find_element(By.CSS_SELECTOR, '#APjFqb')
print(input_box)
```

```python
input_box.send_keys('파이썬')
input_box.send_keys(Keys.ENTER)
```
