```python
import requests
import urllib.request as req
from bs4 import BeautifulSoup
import re
```

```python
res = requests.get('https://news.kbs.co.kr/news/pc/view/view.do?ncd=8369499')
print(res.status_code)
print(res.headers)
```

```python
html = res.text
print(html)
```

```python
soup = BeautifulSoup(html, 'html.parser')
soup
```

```python
title1 = soup.find('h4', class_ = 'headline-title')
print(title1)
```
