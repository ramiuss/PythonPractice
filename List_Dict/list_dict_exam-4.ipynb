```python
# 1. 두 리스트의 공통 요소와 차집합 구하기
# 아래 두 리스트에서

# 공통으로 들어있는 값만 담은 리스트

# 첫 번째 리스트에만 있는 값만 담은 리스트
# 를 각각 출력하세요.

a = [1, 2, 3, 4, 5, 6]
b = [4, 5, 6, 7, 8, 9]
# 여기에 코드를 작성하세요
# 예시 결과: 
# 공통: [4, 5, 6]
# 차집합: [1, 2, 3]

print(set(a) & set(b))
print(set(a) - (set(a) & set(b)))
```

```python
# 2. 문자열에서 중복된 알파벳만 추출하여 set으로 만들기
# 아래 문자열에서 중복되어 나오는 알파벳만 set으로 만들어 출력하세요.

s = "abracadabra"
# 예시 결과: {'a', 'b', 'r'}

from collections import Counter
result = Counter(s)
print(result)

result = [i for i in set(s) if s.count(i) >1]
set(result)
```

```python
# 3. 학생별로 선택한 과목 집합 비교
# 아래와 같이 학생별로 선택한 과목이 딕셔너리로 주어졌을 때,
# 모든 학생이 공통으로 선택한 과목의 set을 구하세요.

subjects = {
    'Tom': {'math', 'english', 'science'},
    'Jane': {'english', 'science', 'art'},
    'Mike': {'science', 'english', 'music'}
}
# 예시 결과: {'english', 'science'}

# tom = list(subjects.values())[0]
# print(tom)
# jane = list(subjects.values())[1]
# print(jane)
# mike = list(subjects.values())[2]
# print(mike)

# 위처럼 복잡하게 하지 말고 직접 키 값을 가져오면 된다...

tom = set(subjects['Tom'])
jane = set(subjects['Jane'])
mike = set(subjects['Mike'])
result = ((tom & jane) & mike)
print(result)
    
```

```python
# 모든 value를 가져와 교집합 계산
common_subjects = set.intersection(*subjects.values())
print(common_subjects)  # {'english', 'science'}
# * 는 리스트나 튜플, 집합 등에서 요소를 하나씩 풀어서 함수에 전달 할 때 사용한다...즉 아래와 같은 뜻이다.

set.intersection({'math', 'english', 'science'}, {'english', 'science', 'art'}, {'science', 'english', 'music'})
```

```python
# 4. 두 딕셔너리의 key와 value 비교
# 아래 두 딕셔너리에서

# key가 모두 같은지 여부 (True/False)

# value가 모두 같은지 여부 (True/False)
# 를 출력하세요.

d1 = {'a': 1, 'b': 2, 'c': 3}
d2 = {'b': 2, 'a': 1, 'c': 4}
# 예시 결과: 
# key 동일: True
# value 동일: False

if set(d1.keys()) == set(d2.keys()):
    print(f'key 는 동일 : {True}')
else :
    print(f'key 는 동일 : {False}')

if set(d1.values()) == set(d2.values()) :
    print(f'value 동일 : {True}')
else :
    print(f'valus 동일 : {False}')

# 아래와 같이 간단하게 한 줄로 할 수 있다. 
key_same = set(d1.keys()) == set(d2.keys()) 
print(key_same)

```

```python
# 5. 리스트에서 등장 횟수가 가장 많은 값 찾기
# 아래 리스트에서 가장 많이 등장하는 값(들)을 set으로 만들어 출력하세요.
# (가장 많이 등장하는 값이 여러 개일 수도 있음)

nums = [1, 2, 2, 3, 3, 3, 4, 4, 4, 5]
# 예시 결과: {3, 4}

from collections import Counter
count = Counter(nums)


result = []
max_count = max(count.values())
for k, v in count.items() :
    if v == max_count :
        result.append(k)
print(result)


```

```python
# 1. 함수에 리스트의 값 언패킹해서 전달하기
# 아래 리스트의 값을 모두 더하는 함수 add(a, b, c, d)가 있습니다.
# 리스트 nums를 이용해 함수에 값을 한 번에 전달해 결과를 출력하세요.

def add(a, b, c, d):
    return a + b + c + d

nums = [10, 20, 30, 40]
# 여기에 코드를 작성하세요
# 예시 출력: 100

print(add(*nums))
```

```python
# 2. 여러 리스트를 하나의 리스트로 합치기 (*와 리스트 언패킹)
# 아래의 세 리스트를 하나의 리스트로 합쳐서 출력하세요.

a = [1, 2]
b = [3, 4]
c = [5, 6]
# 예시 출력: [1, 2, 3, 4, 5, 6]

print(a + b + c)
print([*a, *b, *c])

result = []
result.extend(a)
result.extend(b)
result.extend(c)
print(result)

# 조금 다르게
def add_items(x, y, z) :
    result = []
    for i in (x, y, z) :
        result.extend(i)
    return result
print(add_items(a, b, c))

#리스트를 리스트 통째로 합칠 경우에는 
def add_list(x, y, z) :
    result =[]
    for i in (x, y, z) :
        result.append(i)
    return result
print(add_list(a, b, c))
```

```python
# 3. 함수에서 가변 인자 받기 (*args)
# 여러 개의 숫자를 받아 모두 곱해서 반환하는 함수 multiply를 완성하세요.

def multiply(*args):
    result = 1
    for i in args :
        result *= i
    return result


print(multiply(2, 3, 4))  # 예시 출력: 24
print(multiply(5, 10))    # 예시 출력: 50
```

```python
# 4. 딕셔너리 여러 개를 합쳐서 하나의 딕셔너리 만들기 (** 딕셔너리 언패킹)
# 아래 두 딕셔너리를 언패킹 연산자를 사용해 하나의 딕셔너리로 합치고 출력하세요.

d1 = {'a': 1, 'b': 2}
d2 = {'c': 3, 'd': 4}
# 예시 출력: {'a': 1, 'b': 2, 'c': 3, 'd': 4}

print({**d1, **d2})

```

```python
# 5. 리스트의 일부 요소만 변수에 언패킹해서 할당하기
# 아래 리스트에서 첫 번째 값은 a, 마지막 값은 b, 나머지는 모두 rest 리스트에 할당하세요.

nums = [10, 20, 30, 40, 50]
# 예시 결과:
# a = 10
# rest = [20, 30, 40]
# b = 50

a, *rest, b = nums
print(f'a = {a}')      # 10
print(f'rest = {rest}')   # [20, 30, 40]
print(f'b = {b}')      # 50


```
