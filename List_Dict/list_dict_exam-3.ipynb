```python
# 1. 리스트에서 특정 값 추가/삭제/정렬하기
# 아래 코드에서 리스트에 10을 추가하고, 3을 삭제한 뒤, 내림차순으로 정렬한 결과를 출력하세요.

numbers = [5, 3, 8, 1]
# 여기에 코드를 작성하세요

numbers.append(10)
numbers.remove(3)
print(sorted(numbers, reverse=True))
```

```python
# 2. 딕셔너리 값 조회 및 수정
# 아래의 딕셔너리에서 'math' 점수를 출력하고, 'english' 점수를 100으로 수정한 뒤, 전체 딕셔너리를 출력하세요.

scores = {'korean': 90, 'math': 85, 'english': 95}
# 여기에 코드를 작성하세요
print(scores)

print(scores['math'])
scores['english'] = 100
print(scores)
```

```python
# 3. 리스트와 딕셔너리 변환
# 아래의 딕셔너리를 리스트로 변환하여 (key, value) 쌍의 튜플로 이루어진 리스트를 만들어 출력하세요.

person = {'name': 'Tom', 'age': 20, 'city': 'Seoul'}
# 여기에 코드를 작성하세요

result = list(person.items())
print(result)
```

```python
# 4. 딕셔너리에서 key/value 전체 출력
# # 아래 딕셔너리에서 모든 key와 value를 각각 한 줄에 출력하세요.

fruits = {'apple': 3, 'banana': 5, 'orange': 2}
# 여기에 코드를 작성하세요
for k, v in fruits.items() :
    print(k, v)
```

```python
# 1. 중복 제거 후 정렬하기
# 아래 리스트에서 중복된 값을 제거하고, 남은 값들을 오름차순으로 정렬한 결과를 출력하세요

nums = [4, 2, 7, 4, 9, 2, 1, 7, 3]
# 여기에 코드를 작성하세요
# print(nums)

a = []
for i in nums :
    if i not in a :
        a.append(i)
    else :
        continue
print(sorted(a))
```

```python
# 2. 딕셔너리에서 최대값의 key 찾기
# 아래 딕셔너리에서 값이 가장 큰 key를 찾아 출력하세요.

scores = {'Alice': 88, 'Bob': 95, 'Charlie': 90, 'David': 87}
# 여기에 코드를 작성하세요

for k, v in scores.items() :
    if v == max(scores.values()) :
        print(k)
```

```python
# 3. 리스트의 각 요소를 제곱하여 새로운 딕셔너리 만들기
# 아래 리스트의 각 요소를 key로 하고, 그 제곱을 value로 하는 딕셔너리를 만들어 출력하세요.

nums = [1, 2, 3, 4, 5]
# 예시 결과: {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}

result = {}
for i in nums :
    result[i] = i ** 2
print(result)
```

```python
# 4. 두 리스트를 이용해 딕셔너리 만들기
# 아래 두 리스트를 이용해, 첫 번째 리스트는 key, 두 번째 리스트는 value가 되도록 딕셔너리를 만들어 출력하세요.

keys = ['name', 'age', 'city']
values = ['Jane', 30, 'Busan']
# 예시 결과: {'name': 'Jane', 'age': 30, 'city': 'Busan'}

combine = dict(zip(keys, values))
print(combine)
```

```python
# 5. 딕셔너리에서 value 기준 정렬
# 아래 딕셔너리를 value(점수) 기준으로 내림차순 정렬하여, 정렬된 (이름, 점수) 쌍을 리스트로 출력하세요.

scores = {'Tom': 70, 'Jerry': 85, 'Mike': 60, 'Anna': 95}
# 예시 결과: [('Anna', 95), ('Jerry', 85), ('Tom', 70), ('Mike', 60)]

a = sorted(scores.values())
print(a)

result = sorted(scores, key = lambda x : scores[x])
print(result)

result = sorted(scores.items(), key = lambda x : x[1], reverse = True)
print(result)
```

```python
# 1. 리스트에서 특정 값의 개수 세기
# 아래 리스트에서 숫자 7이 몇 번 등장하는지 세어 출력하세요.

nums = [7, 2, 7, 4, 7, 9, 2, 1, 7, 3]
# 여기에 코드를 작성하세요

count = 0
for i in nums :
    if i == 7 :
        count += 1
print(count)
```

```python
# 2. 딕셔너리에서 value가 80 이상인 key만 리스트로 만들기
# 아래 딕셔너리에서 점수가 80 이상인 학생 이름만 리스트로 만들어 출력하세요.

scores = {'Alice': 78, 'Bob': 95, 'Charlie': 82, 'David': 67, 'Eve': 90}
# 여기에 코드를 작성하세요

result = []
for k, v in scores.items() :
    if v >= 80 :
        result.append(k)
print(result)
```

```python
# 3. 리스트의 각 요소를 문자열로 변환하여 딕셔너리 만들기
# 아래 리스트의 각 요소를 key로 하고, value는 해당 숫자를 문자열로 변환한 딕셔너리를 만들어 출력하세요.

nums = [10, 20, 30, 40]
# 예시 결과: {10: '10', 20: '20', 30: '30', 40: '40'}

result = {}
for i in nums :
    result[i] = str(i)
print(result)

```

```python
# 4. 딕셔너리의 value 총합 구하기
# 아래 딕셔너리에서 모든 value의 합을 구해 출력하세요.

points = {'A': 10, 'B': 20, 'C': 30, 'D': 40}
# 여기에 코드를 작성하세요

sum_value = 0
for k, v in points.items() :
    sum_value += v
print(sum_value)
```

```python
# 5. 리스트에서 중복된 값만 추출하여 새로운 리스트 만들기
# 아래 리스트에서 중복되어 등장하는 값만 골라, 중복 없이 한 번씩만 담은 리스트를 만들어 출력하세요.

nums = [1, 2, 3, 2, 4, 5, 3, 6, 7, 2, 3]
# 예시 결과: [2, 3]

result = []
temp = []
for i in nums :
    if i not in temp :
        temp.append(i)
    else :
        if i not in result :
            result.append(i)
        else :
            continue
print(result)

result = {}
result_list = []
for i in nums :
    if i not in result :
        result[i] = 1
    else :
        result[i] += 1
print(result)

for k, v in result.items() :
    if v > 1 :
        result_list.append(k)
print(result_list)
```

```python
nums = [1, 2, 3, 2, 4, 5, 3, 6, 7, 2, 3]

#count()를 사용하면 반복 횟수를 찾아낼수 있다..
print(nums.count(2)) 

#ser{} 를 사용하면 중복 없이 만들어준다..
result = [x for x in set(nums) if nums.count(x) > 1]
print(result)  # [2, 3]
print(list(x for x in set(nums) if nums.count(x) > 1))
print([x for x in set(nums) if nums.count(x) > 1])

from collections import Counter
counter = Counter(nums)
result = [x for x , cnt in counter.items() if cnt > 1]
print(result)
```
