```python
# 1. 메모이제이션을 사용한 피보나치 최적화
# 재귀를 사용하되 메모이제이션 기법을 적용하여 피보나치 수열의 n번째 항을 효율적으로 계산하는 함수를 작성하세요.

memo = {1 : 1, 2 : 1}
def fibo(n) :
    if n in memo :
        return memo[n]
    else :
        memo[n] = fibo(n-1) + fibo(n-2)
        return memo[n]
    
print(fibo(50))

```

```python
# 2. 데코레이터로 함수 실행 시간 측정
# 함수의 실행 시간을 초 단위로 측정하여 출력하는 데코레이터를 작성하세요.

from datetime import datetime

def spent_time(function) :
    start_time = datetime.now()
    function()
    end_time = datetime.now()
    print(f"함수 실행 시간 : {end_time - start_time} 초")
    
def fibonaci(n) :

    if n == 1 or n ==2 :
        return 1    
    else :
        return fibonaci(n-1) + fibonaci(n-2)
    
def wrapper() :
    fibonaci(35)
    
spent_time(lambda : fibonaci(35))
spent_time(wrapper)

```
