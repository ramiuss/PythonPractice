```python
html_doc = """
<html>
<head>
    <title> BeautifulSoup Example</title>
</head>
<body>
    <h1>Web Crawling Process</h1>
    <p>1. reauest : 웹페이지의 html을 가져오는 역활</p>
    <p>2. BeautifulSuup : html 문서 파싱을 도와준다</p>
    <p>3. 원하는 요소를 선택한다.</p>
</body>
</html>
"""
```

```python
soup = BeautifulSoup(html_doc, 'html.parser')
```

```python
title = soup.title
heading1 = soup.h1
paragraph = soup.p
paragraph_all = soup.find_all('p')
```
