```python
for i in range(1, 5) :
    url = f'https://startcoding.pythonanywhere.com/basic?page={i}'
    res = requests.get(url)
    html = res.text
    soup = BeautifulSoup(html, 'html.parser')
    items = soup.select('.product')
    result = []
    for item in items :
        카테고리 = item.select_one('.product-category').text
        상품명 = item.select_one('.product-name').text
        링크사이트 = item.select_one('.product-name a').attrs['href']
        상품가격 = item.select_one('.product-price').text.split('원')[0]
        result.append((카테고리, 상품명, 링크사이트, 상품가격))

print(result)
```

```python
df = pd.DataFrame(result, columns=[카테고리, 상품명, 링크사이트, 상품가격])
df
```
