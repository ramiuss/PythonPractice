```python
# 1. 문자열 리스트에서 숫자 포함 단어만 소문자로 변환
# 문자열 리스트 words에서, 숫자가 하나라도 포함된 단어만 filter와 lambda로 추출하고, map과 lambda로 소문자로 변환한 리스트를 만드세요.

words = ['Hello1', 'world', 'Python3', 'lambda', 'Map2Filter']
# 결과: ['hello1', 'python3', 'map2filter']

result = []

def find_words(data) :
    for i in data :
        for j in i :
            if j.isdigit() == True :
                result.append(i.lower())
    return result
print(find_words(words))

num_words = list(filter(lambda x : any(i.isdigit() for i in x), words))
print(num_words)
result = list(map(lambda x : x.lower(), num_words))
print(result)
```

```python
# 2. 튜플 리스트에서 두 값의 합이 10 이상인 튜플만 추출
# 아래와 같은 튜플 리스트 data에서, 두 값의 합이 10 이상인 튜플만 filter와 lambda로 추출하세요.

data = [(3, 7), (2, 4), (6, 5), (10, 1), (4, 4)]
# 결과: [(3, 7), (6, 5), (10, 1)]

print(data[0][0], data[0][1])

def sum_tuple(data) :
    result = []
    for i in data :
        if i[0] + i[1] >= 10 :
            result.append(i)
    return result
print(sum_tuple(data))

print(list(map(lambda x : x , list(filter(lambda x : x[0] + x[1] >= 10, data)))))
```

```python
# 3. map, filter로 짝수의 제곱만 추출
# 정수 리스트 numbers에서, filter와 lambda로 짝수만 추출하고, map과 lambda로 제곱값만 리스트로 만드세요.

numbers = [1, 2, 3, 4, 5, 6, 7, 8]
# 결과: [4, 16, 36, 64]

def power_even(data) :
    result = []
    for i in data :
        if i % 2 == 0 :
            result.append(i ** 2)
    return result
print(power_even(numbers))

print(list(map(lambda x : x** 2 , list(filter(lambda x : x % 2 == 0, numbers)))))
```

```python
# 4. 문자열 리스트에서 모음(a, e, i, o, u)으로 시작하는 단어만 대문자로 변환
# 문자열 리스트 words에서, filter와 lambda로 모음으로 시작하는 단어만 추출하고, map과 lambda로 대문자로 변환하세요.

words = ['apple', 'banana', 'orange', 'umbrella', 'grape', 'egg']
# 결과: ['APPLE', 'ORANGE', 'UMBRELLA', 'EGG']

print(words[0][0])

def capitalize_vowels(data) :
    result = []
    for i in data :
        if i[0] in 'a,e,i,o,u' :
            result.append(i[0].upper() + i[1:])
    return result
print(capitalize_vowels(words))

print(list(map(lambda x : x.upper(), list(filter(lambda x : x[0] in 'a,e,i,o,u', words)))))
```

```python
words = ['apple', 'banana', 'orange', 'umbrella', 'grape', 'egg']

def capitalize_word(data) :
    result = []
    for i in data :
        if i[0] in 'a,e,i,o,u' :
            result.append(i.upper())
    return result
print(capitalize_word(words))

print(list(map(lambda x : x.upper(), filter(lambda x : x[0] in 'a, e, u, i, o', words))))
```
