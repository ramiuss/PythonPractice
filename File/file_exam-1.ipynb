```python
import random

hangle = list('가나다라마바사아자차카타파하')

with open('human.csv', 'a', encoding = 'utf-8') as file :

    for i in range(1000) :

        name = random.choice(hangle) + random.choice(hangle) + random.choice(hangle)
        age = random.randrange(20, 100)
        height = random.randrange(150, 200)
        weight = random.randrange(50, 100)
        BMI = weight / (height/100) ** 2
        decision = ''
        if BMI > 25 :
            decision = 'heavy weight'
        elif BMI > 18 :
            decision = 'normal'
        else :
            decision = 'light weight'
        file.write(f'{name}, {age}, {height}, {weight}, {BMI}, {decision}' + '\n')
```
