```python
import time

start_time = time.time()

def factorial(n) :
    result = 1
    for i in range(1, n + 1) :
        result *= i
    return result

print(factorial(40))

end_time = time.time()
print(f'Excution time : {end_time - start_time}')
```

```python
memo = {}
memo[0] = 1

def factorial(n) :
    if n in memo :
        return memo[n]
    memo[n] = n * factorial(n - 1)
    print(memo)
    return memo[n]

print(factorial(5))


```

```python
# 가위  보자기....

n = 100
result = []
for i in range(n + 1) :
    count = 0
    for j in range(n + 1) :
        k = n - (i + j)
        if k >= 0 :
            # print(i, j, k)
            result.append((i, j, k))

count = 0 
for m in result :
    # print(i[0], i[1], i[2])
    if m[0] > 50 or m[1] > 50 or m[2] > 50 :
        print(m[0], m[1], m[2])
        count += 1
        

print(count)          # 위 결과는 40, 30, 30 일 경우가 빠졌으므로 틀린 코드이다...
        
```

```python
n = 4

result = []
for i in range(n + 1) :
    for j in range(n + 1) :
        k = n - (i + j)
        if k >= 0 :
            print(i, j, k)
            result.append((i, j, k))

print(result)

# 이제 각 요소 속에 단독으로 제일 많은 숫자가 있는 경우를 찾아낸다..

count = 0
for i in result :
    if i[0] > i[1] and i[0] > i[2] :
        print(f'가위 {i[0]}, 바위 {i[1]}, 보{i[2]} .가위가 승자 입니다')
        count += 1
    if i[1] > i[2] and i[1] > i[0] :
        print(f'가위 {i[0]}, 바위 {i[1]}, 보{i[2]} .바위가 승자 입니다')
        count += 1
    if i[2] > i[0] and i[2] > i[1] :
        print(f'가위 {i[0]}, 바위 {i[1]}, 보{i[2]} .보자기가 승자 입니다')
        count += 1

print(count)
    
```
