```python
# 문제 1. 문자열 압축 (난이도: 중)
# 주어진 문자열에서 연속으로 나타나는 문자를 압축해서 표현하세요.
# 예를 들어, 문자열이 "aaabbccccdaa"라면 "a3b2c4d1a2"로 압축됩니다.

# 입력 예시
# "aaabbccccdaa"

# 출력 예시
# "a3b2c4d1a2"
```

```python
a = "aaabbccccdaa"
print(a, type(a), a[0])

def compress_data(data) :
    current_chr = data[0]
    count = 0
    result = []
    for i in data :
        if i == current_chr :
            count += 1
        else :
            result.append(current_chr + str(count))
            count = 1
            current_chr = i
            print(result)
    return ''.join(result)

compress_data(a)            # 결과를 보면 마지막이 반영되어 있지 않다..
```

```python
# 문제 2. 두 리스트의 교집합 (난이도: 하)
# 두 개의 정수 리스트가 주어졌을 때, 두 리스트에 모두 존재하는 원소만을 담은 리스트를 반환하세요. 
# 단, 결과는 중복 없이 오름차순으로 정렬되어야 합니다.

# 입력 예시
# list1 = [1, 2, 3, 4, 5, 6, 7]
# list2 = [4, 5, 6, 7, 8, 9, 10]
```

```python
list1 = [1, 2, 3, 4, 5, 6, 7]
list2 = [4, 5, 6, 7, 8, 9, 10]

set_list1 = set(list1)
set_list2 = set(list2)
print(set_list1, set_list2)

print(set_list1 & set_list2)     # 교집합
print(set_list1 | set_list2)     # 합집합
print(set_list1 - set_list2)     # 1 -2 f로 차집합
```

```python
# 1부터 20까지의 숫자가 적힌 카드가 일렬로 놓여 있습니다.
# 처음에는 모든 카드가 뒷면(숫자가 안 보이는 면)으로 놓여 있습니다.

# 이번에는 1번부터 20번까지의 사람이 차례로 나와서 다음과 같이 행동합니다.

# 각 사람은 자신의 번호와 "서로소(공약수가 1)"인 카드만 뒤집습니다.

# 예를 들어, 6번 사람은 1, 5, 7, 11, 13, 17, 19번 카드만 뒤집습니다.
# (6과 서로소인 숫자만 뒤집음)

# 모든 사람이 다 지나간 후,
# 앞면(숫자가 보이는 면)이 보이는 카드의 번호는 무엇일까요?
```

```python
# 서로소 를 찾기 위해 math 라이브러리를 이용해서 gcd
import math

n = 20
result = {}
for card in range(1, n + 1) :
    count = 0
    for person in range( 1, n + 1) :
        if math.gcd(card, person) == 1 :
            count += 1
            result[card] = count
print(result)

for k, v in result.items() :
    if v % 2 == 1 :
        print(k, end=', ')
```

```python
n = 20

def gcd(a, b) :
    while b != 0 :
        a, b = b, a % b
    return a

front_card = {}

for card in range(1, 20 + 1) :
    count = 0
    for person in range(1, n + 1) :
        if gcd(card, person) == 1 :
            count += 1
            front_card[card] = count

print(front_card)

for k, v in front_card.items() :
    if v % 2 == 1 :
        print(k, end=', ')
```

```python
import math

n = 20
front_cards = []

for card in range(1, n + 1):
    count = 0
    for person in range(1, n + 1):
        if math.gcd(card, person) == 1:  # 서로소 판별
            count += 1
    if count % 2 == 1:  # 홀수 번 뒤집힌 카드만 앞면
        front_cards.append(card)

print("앞면이 보이는 카드 번호:", front_cards)
```
