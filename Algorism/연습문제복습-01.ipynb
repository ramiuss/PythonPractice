```python
#위의 결과를 보면 마지막 문자가 포함이 않되어 있다...

c = "aaabbccccdaa"

def compress_str(a) :
    cur_chr = a[0]
    count = 1
    result = []

    for i in a[1::] :
        if i == cur_chr :
            count += 1
        else :
            result.append(cur_chr + str(count))
            count = 1
            cur_chr = i

    result.append(cur_chr + str(count))
    return result

print(''.join(compress_str(c)))


```

```python
n = 4 

count = 0
for s in range(n + 1) :
    for r in range(n - s + 1) :
        p = n - (r + s) 
        print(s, r, p)
        result = [s, r, p]
        if result.count(max(result)) == 1 :
            count += 1

print(count)
```

```python
n = 100

count = 0
for i in range(0, n + 1) :
    for j in range (i , n + 1) :
        all = [i, j - i, n - j]
        if all.count(max(all)) == 1 :
            count += 1

print(count)
```
